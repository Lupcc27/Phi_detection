```python
#change pixels to image
import numpy as np
from PIL import Image

def pixels_to_image(pixels, width, height):
    pixels = np.array(pixels, dtype=np.uint8)
    pixels = pixels.reshape((height, width))
    image = Image.fromarray(pixels, 'L')  # 'L' mode for grayscale
    return image
```

```python
import pandas as pd
data_path = "C:/Users/andre/.cache/kagglehub/datasets/davilsena/ckdataset/versions/2/ckextended.csv"
df = pd.read_csv(data_path)
df.describe()
```

## Division en carpetas por emocion 

```python
new_path = "C:/Users/andre/.cache/kagglehub/datasets/davilsena/ckdataset/versions/2/ckextended.csv"
for index, row in df.iterrows():
    pixels = list(map(int, row['pixels'].split()))
    image = pixels_to_image(pixels, 48, 48)
    if row['emotion'] == 0:  # Check if emotion is 'happy'
        image.save(f"ck_dataset/0/image_{index}.png")
    elif row['emotion'] == 1:  # Check if emotion is 'sad'
        image.save(f"ck_dataset/1/image_{index}.png")
    elif row['emotion'] == 2:  # Check if emotion is 'angry'
        image.save(f"ck_dataset/2/image_{index}.png")
    elif row['emotion'] == 3:  # Check if emotion is 'surprised'
        image.save(f"ck_dataset/3/image_{index}.png")
    elif row['emotion'] == 4:  # Check if emotion is 'neutral'
        image.save(f"ck_dataset/4/image_{index}.png")
    elif row['emotion'] == 5:  # Check if emotion is 'disgust'
        image.save(f"ck_dataset/5/image_{index}.png")
    elif row['emotion'] == 6:  # Check if emotion is 'fear'
        image.save(f"ck_dataset/6/image_{index}.png")
    else:
        image.save(f"ck_dataset/7/image_{index}.png")
print("Images saved successfully.")
```
